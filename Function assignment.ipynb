```python
#Question 4
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

primes = [n for n in range(1, 201) if is_prime(n)]
print(primes)
```

```python
#Question 5
class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.b
```

```python
#Question 6
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i
```

```python
#Question 7
def read_file(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()
```

```python
#Question 10
def remove_vowels(s):
    vowels = 'aeiouAEIOU'
    return ''.join(filter(lambda x: x not in vowels, s))

print(remove_vowels("Hello World"))
```

```python
#Question 11
orders = [
    ['Order-001', 5, 15.0],
    ['Order-002', 2, 45.0],
    ['Order-003', 1, 80.0],
]

result = list(map(lambda order:
                  (order[0], order[1] * order[2] if order[1] * order[2] >= 100 else order[1] * order[2] + 10),
                  orders))

print(result)
```

1. Difference between a function and a method in Python?
   - A function is a block of code that performs a specific task and is defined using def.
   - A method is a function that is associated with an object (belongs to a class).
   - Example:
       # Function
      def greet():
    return "Hello"

        # Method (attached to a string object)
         "hello".upper()

2. Function arguments and parameters?
   - Parameter: The variable listed inside the parentheses in the function definition.
   - Argument: The value passed to the function when it is called.
Example:
      def add(x, y):  # x and y are parameters
       return x + y
       add(3, 4)  # 3 and 4 are arguments

3. Different ways to define and call a function:
  - Normal function definition
  - Lambda function
  - Function using def inside another function (nested)
Example:
# Normal function
def square(x):
    return x * x

# Lambda function
square_lambda = lambda x: x * x

# Calling
print(square(5))
print(square_lambda(5))

4. Purpose of the return statement?
 - Return is used to send the result from a function back to the caller.
Example:
def multiply(a, b):
    return a * b

5. Iterators vs Iterables?

   - Iterable: Object you can loop over (like list, tuple, string).
   - Iterator: Object that produces next values using __next__().
Example:
iterable = [1, 2, 3]
iterator = iter(iterable)
print(next(iterator))  # 1    

6. Concept of generators?
Generators are functions that yield values one at a time using yield.
Example:
def gen_numbers():
    yield 1
    yield 2
    yield 3

7. Advantages of using generators:
   - Memory efficient (don't store entire sequence in memory).
   - Faster for large datasets.  

8. Lambda function and its use?
   - Anonymous (unnamed) small function.
   - Used when a small function is needed for a short period.
Example:
square = lambda x: x * x
print(square(5))

9.Purpose and usage of map() function?

  Example:
nums = [1, 2, 3, 4]
